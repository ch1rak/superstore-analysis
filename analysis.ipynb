```python
import sqlite3
conn = sqlite3.connect(":memory:")
df.to_sql("orders", conn, index=False)
```

```python
query = """
SELECT Region, ROUND(SUM(Sales), 0) AS sales, ROUND(SUM(Profit), 0) AS profit
FROM orders
GROUP BY Region
ORDER BY profit DESC
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT "Customer Name" , sum(Sales) AS total
FROM orders 
GROUP BY "Customer Name"
ORDER BY total DESC
LIMIT 5
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT strftime('%Y' , "Order Date") AS Year , 
COUNT(DISTINCT "Order ID") AS orders
FROM orders
GROUP BY Year
ORDER BY Year
"""
pd.read_sql(query , conn)
```

```python
query = """
select "Sub-Category" , 
sum(Profit) AS profit 
from orders
group BY "Sub-category" 
having profit < 0
order BY profit"""
pd.read_sql(query , conn)
```

```python
query = """
select "Segment" ,
AVG("Discount") AS discount
from orders
group by "Segment"
order by discount"""
pd.read_sql(query , conn)
```

```python
last_date = df["Order Date"].max() # последний день в данных
rfm = df.groupby("Customer ID").agg(       # для каждого клиента считаем три числа
    recency=("Order Date", lambda d: (last_date - d.max()).days),  # дней с последней покупки
    frequency=("Order ID", "nunique"),     # сколько разных заказов
    monetary=("Sales", "sum"),             # на какую сумму купил
)
rfm.head()
```

RFM - таблица

```python
from sklearn.preprocessing import StandardScaler
X = StandardScaler().fit_transform(rfm) #fit запоминает среднее и разброс каждого столбца, transform применяет формулу.
```

масштабирование признаков

```python
from sklearn.cluster import KMeans
inertias = []
for k in range (2 , 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    inertias.append(km.inertia_)  # насколько точки далеки от своих центров
plt.plot(range(2, 11), inertias, marker="o")
plt.xlabel("k"); plt.ylabel("inertia"); plt.show()
    
```

метод локтя


```python
km = KMeans(n_clusters=4, random_state=42, n_init=10) #создание модели
rfm["cluster"] = km.fit_predict(X)    # fit (обучение: алгоритм находит 4 итоговых центра)  predict (для каждого клиента говорит, к какому центру он ближе, то есть номер кластера.)
rfm["cluster"].value_counts()          # сколько клиентов в каждом кластере
```

```python
rfm.groupby("cluster").mean().round(1)
```

портрет клиента из каждой группы

```python
summary = rfm.groupby("cluster").agg(
    clients=("recency", "count"),        # сколько клиентов в группе
    recency=("recency", "mean"),         # средние R, F, M
    frequency=("frequency", "mean"),
    monetary=("monetary", "mean"),
).round(1).sort_values("monetary", ascending=False)
summary
```
